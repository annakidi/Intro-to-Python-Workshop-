# Concept 1: Input, Output, Variables, and Math
To make a program interactive, we need to ask the user for information (input), store that information (variables), change its data type so we can do math with it (casting), and show the results (print).
input() always captures text (a string). If you want to do math with a number, you must convert it using int() for whole numbers or float() for decimals.



Python can do maths:

| Symbol | Meaning | Example | Result |
|---|---|---|---|
| `+` | add | `7 + 2` | `9` |
| `-` | subtract | `7 - 2` | `5` |
| `*` | multiply | `7 * 2` | `14` |
| `/` | divide | `7 / 2` | `3.5` |
| `**` | power | `2 ** 3` | `8` |

Python follows the usual order of operations: powers first, then `*` and `/`, then `+` and `-`. Use brackets to take control: `(1 + 2) * 3`.




**Example: Age Next Year Calculator**

````python
user_age_text = input("How old are you? ")
user_age_number = int(user_age_text)
next_year_age = user_age_number + 1
print("Next year, you will be:", next_year_age)
````

### Understanding f-strings

You might notice an `f` before some strings in Python, like `f"You solved {word}."` This is called an **f-string** (short for formatted string literal). It is a convenient way to insert variables or expressions directly inside a string by wrapping them in curly braces `{}`.

**Example:**

````python
name = "Alice"
score = 95
print(f"Great job, {name}! Your score is {score}.")
````

# 2: Conditionals (Decision Making)
Programs need to make decisions based on different situations. We use if, elif (else if), and else statements to run specific blocks of code only when certain conditions are true.
You can check conditions using operators like > (greater than), < (less than), == (equal to), or combine them using and.

**Example: Simple Grading System**

````python
score = float(input("Enter your test score: "))

if score >= 90:
    print("You got an A!")
elif score >= 80:
    print("You got a B!")
else:
    print("Keep practicing!")
````

# Concept 3: While Loops

Sometimes we want our code to keep repeating as long as a certain condition remains true. A while loop checks a condition, runs the code inside it, and then checks the condition again.

**Example: Counting down from 5**

````python
timer = 5

while timer > 0:
    print(timer)
    timer = timer - 1 

print("Blastoff!")
````

# Concept 4: For Loops
When you know exactly how many times you want a block of code to repeat, a for loop combined with the range() function is the best tool. range(x) will run the loop x times.

**Example: The Savings Tracker**

````python
bank_balance = 0

# This loop runs exactly 4 times
for week in range(4):
    bank_balance = bank_balance + 20
    print("Week complete. Current balance:", bank_balance)
````

# Concept 5: Lists and Importing Modules
Sometimes we need to store multiple pieces of data in one place, like a collection of words. We use Lists (created with square brackets [] and can be changed after it's created) or Tuples (created with parentheses () and cannot be changed) for this.
Python also has built-in tools that aren't loaded by default. We use the import keyword to bring them into our program. For example, the random module lets us pick random items from a list.

**Example: The Random Snack Picker**

````python
from random import choice

# A tuple holding multiple strings
snacks = ("apple", "chips", "cookie", "carrots")

# Pick a random item from the tuple
my_snack = choice(snacks)
print("You should eat a:", my_snack)
````
